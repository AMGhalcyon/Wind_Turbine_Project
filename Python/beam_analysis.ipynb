```python
#import
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.set()
```

```python
#defining parameters
L = 1.2
b = 0.04
h = 0.08

E_aluminium = 69e9
E_fiberglass = 35e9

w = 100
```

```python
# Moment of Inertia 

I = (b*(h**3))/12

print(I)
```

```python
# Maximum bending moment

M_max = (w*(L**2))/2
# occurring at the root
print(M_max)
```

```python
# Maximum bending stress
c = h/2
sigma_max = (M_max * c ) /I

print(sigma_max)

# so maximum bending stress occurs at root and on the outer surface
```

```python
# computing maximum deflection, which occurs at the tip

delta_max_aluminium = (w*(L**4))/(8*E_aluminium * I)

delta_max_fiberglass = (w*(L**4))/(8*E_fiberglass * I)

print(delta_max_aluminium)

print(delta_max_fiberglass)

# from the formula the tip deflection is very sensitive to Length of the beam
```

```python
# Plotting shear force diagram
# Span positions
x = np.linspace(0, L, 200)

# Shear force equation
V = w * (L - x)

# Plot
plt.figure(figsize=(8,5))
plt.plot(x, V)

plt.xlabel("Blade Span Position (m)")
plt.ylabel("Shear Force (N)")
plt.title("Shear Force Diagram (Cantilever Beam)")
plt.grid(True)

plt.show()
```

```python
# Plotting Moment diagram
# Span positions
x = np.linspace(0, L, 200)

# Moment equation
Moment = w*L*x - (w * (L**2)/2) - (w * x**2/2)

# Plot
plt.figure(figsize=(8,5))
plt.plot(x, Moment)

plt.xlabel("Blade Span Position (m)")
plt.ylabel("Moment (M)")
plt.title("Moment Diagram (Cantilever Beam)")
plt.grid(True)

plt.show()
```

```python
# Plotting of deflection as x progresses using singularity function
# Span positions
x = np.linspace(0, L, 300)

# Deflection equation ( just for aluminium, can substitute for fiberglass too)
y = (w / (24 * E_aluminium * I)) * (x**4 - 4*L*x**3 + 6*(L**2)*x**2)

# Plot
plt.figure(figsize=(8,5))
plt.plot(x, y)

plt.xlabel("Blade Span Position (m)")
plt.ylabel("Deflection (m)")
plt.title("Cantilever Beam Deflection")
plt.grid(True)

plt.show()
```
