```python
import numpy as np
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt
sns.set()
```

```python
#creating simulation matrix using nested loops
blade_lengths = [0.8, 1.2, 1.6]
chord_lengths = [150, 200, 300]
materials     = ["Aluminium", "Fiberglass", "Carbon fiber"]
loads         = [500, 1000, 1500]

material_abb = {'Aluminium': 'AL', "Fiberglass" : 'FBG', 'Carbon fiber': 'CBF' }

rows = []
for L in blade_lengths:
    for C in chord_lengths:
        for M in materials:
            for P in loads:
                sim_id = f"BL{L}_C{C}_{material_abb[M]}_P{P}"
                rows.append({ 'ID': sim_id, 
                             'Blade_Length_m' : L,
                             'Root_Chord_mm' : C,
                             'Material' : M,
                             'Applied_Load_Pa' : P,
                             'Status' : 'Pending',
                             'Max_Deformation_m' : "",
                             'Max_Equiv_Stress_Pa' : ""
                })

blade_data = pd.DataFrame(rows)
```
